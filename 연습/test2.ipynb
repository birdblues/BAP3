```python
import pymc as pm
import numpy as np
import pytensor.tensor as pt
from pytensor.graph.op import Op

# 블랙-숄즈 로그 우도 사용자 정의 연산 클래스 정의
class BlackScholesLogLik(Op):
    # 입력과 출력 타입을 실수 스칼라로 지정
    itypes = [pt.dscalar, pt.dscalar]
    otypes = [pt.dscalar]

    # 실제 연산을 수행하는 메서드
    def perform(self, node, inputs, outputs):
        sigma, r = inputs  # 입력값(변동성, 이자율)
        # 블랙-숄즈 모형의 로그 우도 계산 (여기서는 예시로 단순 계산)
        log_likelihood = -0.5 * ((sigma - 0.2)**2 + (r - 0.05)**2)
        outputs[0][0] = np.array(log_likelihood)  # 결과 반환

# 사용자 정의 연산 인스턴스 생성
log_lik = BlackScholesLogLik()

# PyMC 모델 정의
with pm.Model():
    # sigma와 r을 Uniform 분포로 정의
    sigma = pm.Uniform('sigma', lower=0.01, upper=1.0)
    r = pm.Uniform('r', lower=0.0, upper=0.1)
    # 사용자 정의 로그 우도를 모델에 추가
    pm.Potential('likelihood', log_lik(sigma, r))
    # 샘플링 실행
    trace = pm.sample(1000)
```
